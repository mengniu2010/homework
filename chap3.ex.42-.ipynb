a) 直接用栈检查，时间O(n),空间O(n)

````python
def is_balance_str(s: str) -> bool:
    stk = []
    for ch in s:
        if ch in "([":
            stk.append(ch)
        else:
            if (stk and ch == ")" and stk[-1] == "(") or (
                stk and ch == "]" and stk[-1] == "["
            ):
                stk.pop()
            else:
                return False
    return len(stk) == 0

print (is_balance_str("()[[[[(())]]]]")==True)
print(is_balance_str("()[[[[([)]]]]")==False)
print(is_balance_str("()[][()]]")==False)
print(is_balance_str("([)]")==False)
print(is_balance_str("")==True)
````

b) 动态规划。 $f(i,j)$表示$s[i:j]$范围内最长平衡子序列


$$
f(i,j) = 
\begin{cases}
0 & i==j \\
0 & i+1 == j , \neg pair(i,j) \\
2 & i+1 == j , pair(i,j) \\
f(i+1, j-1)+1 & pair(i,j) \\
max\{f(i+1,j), f(i,j-1)\} & \neg pair(i,j)
\end{cases}
$$

依赖：
- $i<j$ 

```
(i, j-1) -->     (i,j)
(i+1, j-1)  --> (i,j)
(i+1,j)    -->  (i,j)

```

````python
def balanced_longest_seq(s: str) -> int:
    n = len(s)
    table = [[0] * n for _ in range(n)]

    def pair(i: int, j: int) -> bool:
        ss = "".join([s[i], s[j]])
        return ss == "()" or ss == "[]"

    for i in range(n - 1):
        if pair(i, i + 1):
            table[i][i + 1] = 2
    for d in range(1, n):
        for i in range(n):
            j = i + d
            if j >= n:
                continue
            if pair(i,j):
                table[i][j] = table[i+1][j-1]+2
            else:
                table[i][j] = max(table[i+1][j], table[i][j-1])
    print(tabulate(table, headers=s,showindex=list(s)))
    return table[0][n-1]


print(balanced_longest_seq("([])[]")) #14
````
